```python
df = pandas.DataFrame(filtered_data)
display(df)
```

```python
blacklist = ["askscience", "megathread", "[", "x-post", "included", "/r/", "xpost", "reddit"]

mask = ~df['title'].apply(lambda x: any(keyword in x for keyword in blacklist))
df = df[mask]

mask = df['title'].apply(lambda x: x.endswith('?'))
df = df[mask]



mask = ~df['title_urls.url'].apply(lambda x: len(x) > 0)
df = df[mask]
df = df.reset_index()


# df_filtered['title'] = df_filtered['title'].map(lambda x: x.strip('"').replace('""', '"'))
display(df)

df['title'].to_csv(f'{wd}/ask_science.csv', quoting=csv.QUOTE_NONE, sep="\t")
```
